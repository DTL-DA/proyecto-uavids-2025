# 6. Guía de procedimiento para clasificación con datos tabulares

El enunciado indica cómo proceder según el tipo de problema. La tabla contrasta cada punto de su guía para clasificación tabular con lo que se hace en este trabajo, y señala dónde se desarrolla.

| Aspecto | Guía del enunciado | Aplicación en este trabajo | Dónde |
|---|---|---|---|
| Partición | Estratificada por clase; agrupada si hay varios registros por entidad; validación cruzada anidada si el tamaño es reducido | Estratificada por clase y agrupada por bloques de filas; sin validación anidada, porque hay decenas de miles de flujos | Secciones {ref}`2.1.3 <sec-2-1-3>` y {ref}`4.1 <sec-4-1>` |
| Preprocesamiento | Imputación, escalado y codificación ajustados solo en entrenamiento; *embeddings* para categóricas de alta cardinalidad | No hace falta imputar; transformación y estandarización ajustadas solo en entrenamiento; las categóricas se excluyen, porque funcionan como atajos | Secciones {ref}`2.1.2 <sec-2-1-2>`, {ref}`2.1.4 <sec-2-1-4>` y {ref}`2.2.1 <sec-2-2-1>` |
| Modelos base típicos | MLP superficial, MLP profundo regularizado y red residual tabular | Los que resultan de la tabla de frecuencia de la literatura, como exige el enunciado | Sección {ref}`3.2 <sec-3-2>` |
| Pérdida | Entropía cruzada, ponderada o *focal* ante desbalance | Entropía cruzada ponderada | Secciones {ref}`2.2.5 <sec-2-2-5>` y {ref}`4.2.2 <sec-4-2-2>` |
| Evaluación | F1 macro, AUC-ROC, AUC-PR, MCC y calibración | Las cinco, además de las métricas declaradas en la sección {ref}`1.1 <sec-1-1>` | Secciones {ref}`4.4 <sec-4-4>` y {ref}`5.1 <sec-5-1>` |
| Inferencia | DeLong, McNemar y Friedman con *post-hoc* | Las tres, con Nemenyi y Holm como *post-hoc* | Sección {ref}`5.4 <sec-5-4>` |
| Referencia de contexto | Regresión logística, para dimensionar el margen | Regresión logística con regularización L2, además de un boosting | Secciones {ref}`2.2.3 <sec-2-2-3>` y {ref}`5.3 <sec-5-3>` |
